```python
import cv2
import numpy as np

img = cv2.imread(r"C:\Users\kanha\Documents\saksham bansal\stone paper\image assets\Timothee.jpg")
img = cv2.resize(img, (600, 600))

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
_, th = cv2.threshold(gray, 7, 255, cv2.THRESH_BINARY)

contours, hierarchy = cv2.findContours(
    th, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE
)

# Draw all contours once
cv2.drawContours(img, contours, -1, (255, 0, 0), 3)
ar=[]
# Draw centroid for each contour
for cnt in contours:
    m = cv2.moments(cnt)
    if m["m00"] != 0:
        x = int(m["m10"] / m["m00"])
        y = int(m["m01"] / m["m00"])
        cv2.circle(img, (x, y), 4, (0, 0, 255), -1)
        a=cv2.contourArea(cnt)
        ar.append(a)
        ep=0.1*cv2.arcLength(cnt,True)
        d=cv2.approxPolyDP(cnt,ep,True)
        h=cv2.convexHull(d)
        x,y,w,h=cv2.boundingRect(h)
        cv2.rectangle(img,(x,y),(x+w,y+h),(0,255,0),5)
cv2.imshow("window", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

img = cv2.imread(r"C:\Users\kanha\OneDrive\Pictures\group.png")
template = cv2.imread(r"C:\Users\kanha\OneDrive\Pictures\ansh.png")

gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
gray_template = cv2.cvtColor(template, cv2.COLOR_BGR2GRAY)

orb = cv2.ORB_create(1000)

kp1, des1 = orb.detectAndCompute(gray_template, None)
kp2, des2 = orb.detectAndCompute(gray_img, None)

bf = cv2.BFMatcher(cv2.NORM_HAMMING)

matches = bf.knnMatch(des1, des2, k=2)

# Lowe's ratio test (important)
good = []
for m, n in matches:
    if m.distance < 0.75 * n.distance:
        good.append(m)

print("Good matches:", len(good))

if len(good) > 10:

    src_pts = np.float32([kp1[m.queryIdx].pt for m in good]).reshape(-1,1,2)
    dst_pts = np.float32([kp2[m.trainIdx].pt for m in good]).reshape(-1,1,2)

    M, mask = cv2.findHomography(src_pts, dst_pts, cv2.RANSAC, 5.0)

    h, w = gray_template.shape
    pts = np.float32([[0,0],[0,h],[w,h],[w,0]]).reshape(-1,1,2)
    dst = cv2.perspectiveTransform(pts, M)

    img = cv2.polylines(img, [np.int32(dst)], True, (0,255,0), 3)

else:
    print("Not enough matches found")

cv2.imshow("Detected Person", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
